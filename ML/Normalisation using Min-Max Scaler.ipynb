```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("wine_data.csv",header=None,usecols=[0,1,2])
df
```

```python
df.columns=["Class Label","Alcohol","malic_acid"]
```

```python
df
```

```python
x=df.iloc[:,[1,2]]
x
```

```python
y=df.iloc[:,[0]]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_train_scaled
```

```python
x_train_scaled=pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_train_scaled
```

```python
x_test_scaled=scaler.transform(x_test)
x_test_scaled
```

```python
x_test_scaled=pd.DataFrame(x_test_scaled,columns=x_test.columns)
x_test_scaled
```

```python
np.round(x_train.describe(),1)
```

```python
np.round(x_train_scaled.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))

ax1.set_title("before scaling")
ax1.scatter(x_train["Alcohol"],x_train["malic_acid"],c=y_train.values)


ax2.set_title("after scaling")
ax2.scatter(x_train_scaled["Alcohol"],x_train_scaled["malic_acid"],c=y_train.values)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))

ax1.set_title("before scaling")
sns.kdeplot(x_train["Alcohol"],ax=ax1,label="Alcohol")
sns.kdeplot(x_train["malic_acid"],ax=ax1,label="malic_acid")
ax1.legend()

ax2.set_title("after scaling")
sns.kdeplot(x_train_scaled["Alcohol"],ax=ax2,label="Alcohol")
sns.kdeplot(x_train_scaled["malic_acid"],ax=ax2,label="malic_acid")
ax2.legend()
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))

ax1.set_title("before scaling")
sns.kdeplot(x_train["Alcohol"],ax=ax1,label="Alcohol")

ax1.legend()

ax2.set_title("after scaling")
sns.kdeplot(x_train_scaled["Alcohol"],ax=ax2,label="Alcohol")

ax2.legend()
```
